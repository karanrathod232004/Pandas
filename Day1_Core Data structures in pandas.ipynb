```python
import pandas as pd
```

Series - 1D Labeled array

```python
#Create a series 
marks =pd.Series([99,67,78,89,54,67,82])
print(marks)
```

```python
#Add index name 
marks =pd.Series([99,67,78,89,54,67,82],index=["Karan","shreya","Jashmin","Nik","Jack","Avinash","Rohit"])
print(marks)

print("-"*40)
Top_marks= marks >90
print(marks[Top_marks])
```

DataFrame - 2D labeled table 

```python
data = {
    "Name" : ["Alice","Bob","Charli"],
    "Age" :[25,30,32],
    "City":["Delhi","Mumbai","Banglore"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Q1 Create series 5 student marks 
marks = pd.Series([88,67,19,56,32])
print(marks)
print("-"*40)
Pass_student = marks > 33
Fail_student = marks < 33
print("Pass student :",marks[Pass_student])
print("-"*40)
print("Fail student :",marks[Fail_student])
```
