#clean New_Data 

```python
new_data = pd.read_csv(r"C:\Users\Ridhi Rathore\Downloads\pandas_cleaning_practice_10_rows.csv")
new_data
```

```python
#Convert name columns into title case and fill nan value 
new_data["Name"] = new_data["Name"].fillna("Unknown")
new_data["Name"] = new_data["Name"].str.title()
```

```python
#City name convert to title case and fill nan value 
new_data["City"] = new_data["City"].fillna("Unknown")
new_data["City"] = new_data["City"].str.title()
```

```python
#Department column convert propar formate 
new_data["Department"] = new_data["Department"].str.lower().fillna("Unknown")
new_data["Department"] = new_data["Department"].map({"sales":"Sales","it":"IT","hr":"HR"})
```

```python
#Salary nan value fill
new_data["Salary"] = pd.to_numeric(new_data["Salary"],errors="coerce")
new_data["Salary"] = new_data["Salary"].fillna(new_data["Salary"].mean()).astype(int)
```

```python
#Date convert propar form
new_data["Joining_Date"] = new_data["Joining_Date"].fillna(new_data["Joining_Date"].ffill())
new_data["Joining_Date"] = pd.to_datetime(new_data["Joining_Date"],format="mixed")
new_data["Joining_Date"] = new_data["Joining_Date"].dt.strftime("%d-%m-%Y")
```

```python
#Performance Fill nan value
new_data["Performance"] = new_data["Performance"].fillna(new_data["Performance"].mean())
```

```python
#Add new columns perfromance level 
def Performance_level(x):
    if x >= 4.5:
        return "Excellent"
    elif x >= 4.0:
        return "Good"
    else:
        return "Average"
new_data["Performance_level"] = new_data["Performance"].apply(Performance_level)
```

```python
#Status column convert parfact form
new_data["Status"] = new_data["Status"].fillna("not").str.title()
```
