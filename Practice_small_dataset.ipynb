#Handle all missing value 

```python
#Fill customer missing value 
data["Customer"] = data["Customer"].fillna("Unknown")
```

```python
#Fill missing city name 
data["City"] = data["City"].fillna("Unknwon")
```

```python
#Fill most frequent value of  missing Category 
data["Category"].value_counts()
```

```python
data["Category"] = data["Category"].fillna(data["Category"].mode()[0])
```

```python
#fill missing quantity value 
data["Quantity"] = data["Quantity"].fillna(data["Quantity"].median())
```

```python
#Fill missing discount value 
data["Discount"] = data["Discount"].fillna(0)
```

```python
#Fill missing date 
data["Order_Date"] = data["Order_Date"].ffill()
```

```python
#Fill missing Payment_status 
data["Payment_Status"] = data["Payment_Status"].fillna("Unknown")
```

```python
#Chack all missing value 
data.isnull().sum()   #All missing value fill 
```

```python
#Remove duplicate data 
data.duplicated()
```

```python
#Drop duplicate value 
data = data.drop_duplicates().copy()
```
