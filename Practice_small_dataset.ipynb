#Handle all missing value 

```python
#Fill customer missing value 
data["Customer"] = data["Customer"].fillna("Unknown")
```

```python
#Fill missing city name 
data["City"] = data["City"].fillna("Unknwon")
```

```python
#Fill most frequent value of  missing Category 
data["Category"].value_counts()
```

```python
data["Category"] = data["Category"].fillna(data["Category"].mode()[0])
```

```python
#fill missing quantity value 
data["Quantity"] = data["Quantity"].fillna(data["Quantity"].median())
```

```python
#Fill missing discount value 
data["Discount"] = data["Discount"].fillna(0)
```

```python
#Fill missing date 
data["Order_Date"] = data["Order_Date"].ffill()
```

```python
#Fill missing Payment_status 
data["Payment_Status"] = data["Payment_Status"].fillna("Unknown")
```

```python
#Chack all missing value 
data.isnull().sum()   #All missing value fill 
```

```python
#Remove duplicate data 
data.duplicated()
```

```python
#Drop duplicate value 
data = data.drop_duplicates().copy()
```

#Remove extra spaces from text columns 

```python
#Remove customer columns extra space 
data["Customer"] = data["Customer"].str.strip()
```

#Convert text columns into proper format

```python
#Customer name columns convert title case
data["Customer"] = data["Customer"].str.title()
```

```python
#City name columns convert title case
data["City"] = data["City"].str.title()
```

```python
#Category name columns convert title case 
data["Category"] = data["Category"].str.title()
```

#Date clean and convert propar date 

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"],format="mixed",dayfirst=True).dt.strftime("%d-%m-%Y")
```

```python
#Quentity column  convert integer form
data["Quantity"] = data["Quantity"].astype(int)
```

```python
#Create order_size 
def order_size(x):
    if x >= 50000:
        return "Big order"
    else:
        return "Normal order"
data["Order_size"] = data["Total_Amount"].apply(order_size)
```
