#Data cleaning preprocessing 

```python
#Fill missing customer name 
data = data.copy()
data["Customer"] = data["Customer"].fillna("Unknown")
data
```

```python
#change electronics to Electronics
data["Category"] = data["Category"].replace({"electronics":"Electronics"})
data
```

```python
#Order data convert datetime 
data["Order_Date"] = pd.to_datetime(data["Order_Date"])
```
