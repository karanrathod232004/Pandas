#Filtering with conditions

```python
df
```

```python
#Whose salary is greaterthan 60000
df[df["Salary"] > 60000]
```

```python
#who employe work IT department 
df[df["Department"] == "IT"]
```

```python
#Whose employees 3 year experience
df[df["Experience"] >=3]
```

```python
#Whose salary less than 50000
df[df["Salary"] < 50000]
```

```python
#Who work in the IT Department and salary is 60000 up
df[(df["Department"] == "IT") & (df["Salary"] >= 60000)]
```

```python
#Who work in the HR or Sales department 
df[(df["Department"] == "HR") | (df["Department"] == "Sales")]
```

```python
#Who have more than 2 year experience and salary 55000 up 
df[(df["Experience"] >= 2) & (df["Salary"] >= 55000)]
```

```python
#Who work in the IT department and have more than 3 years of experience
df[(df["Department"] == "IT" ) & (df["Experience"] >= 3)]
```

```python
#Whose salary is between 50000 and 70000
df[(df["Salary"] > 50000) & (df["Salary"] < 70000)]
```

```python
#Who work in the Finance department and salary is 60000 up
df[(df["Department"] == "Finance") & (df["Salary"] >= 60000)]
```
